```python
import numpy as np
import pandas as pd
import altair as alt
from scipy.stats import chi2

# Fijamos la semilla para reproducibilidad
np.random.seed(1)

# Definimos los grados de libertad de la distribución Chi-cuadrada
grados_libertad = 1

# 1. Generamos 100 datos pseudo-aleatorios de la distribución Chi-cuadrada
datos_chi = np.random.chisquare(df=grados_libertad, size=100)

# Mostramos la media y la varianza de nuestra muestra
print(f"Media muestral: {datos_chi.mean():.4f} (Teórica: {grados_libertad})")
print(f"Varianza muestral: {datos_chi.var():.4f} (Teórica: {2 * grados_libertad})")
```

```python
# 2. Preparar el DataFrame para los datos empíricos
df_empirico_chi = pd.DataFrame({'valores': datos_chi})

# 3. Calcular la densidad de probabilidad teórica
x_teorico_chi = np.linspace(0, np.max(datos_chi) + 2, 100)
y_teorico_chi = chi2.pdf(x_teorico_chi, df=grados_libertad)

df_curva_chi = pd.DataFrame({
    'x': x_teorico_chi,
    'y': y_teorico_chi
})

# 4. Construir el gráfico con Altair
histograma_chi = alt.Chart(df_empirico_chi).mark_bar(opacity=0.6, color='lightgreen').encode(
    alt.X('valores:Q', bin=alt.Bin(maxbins=15), title='Valor'),
    alt.Y('density:Q', title='Densidad')
)

linea_teorica_chi = alt.Chart(df_curva_chi).mark_line(color='darkgreen', size=2).encode(
    x='x:Q',
    y='y:Q'
)

grafico_chi_final = (histograma_chi + linea_teorica_chi).properties(
    title=f'Comparación de Datos Empíricos vs. Distribución Chi-cuadrada (df={grados_libertad})',
    width=600,
    height=400
).interactive()

grafico_chi_final
```

```python
import numpy as np
from scipy import stats

# Pesos de las 10 sandias (en libras)
datos = np.array([7.72, 9.58, 12.38, 7.77, 11.27, 8.80, 11.10, 7.80, 10.17, 6.00])

# Parametros
n = len(datos)
alpha = 0.05
grados_libertad = n - 1

# Varianza muestral
s2 = np.var(datos, ddof=1)

# Valores criticos de chi cuadrado
chi_inferior = stats.chi2.ppf(alpha / 2, grados_libertad)
chi_superior = stats.chi2.ppf(1 - alpha / 2, grados_libertad)

# Intervalo de confianza para la varianza poblacional
limite_inferior = (grados_libertad * s2) / chi_superior
limite_superior = (grados_libertad * s2) / chi_inferior

# Resultados usando f-strings corregidos
print(f"Tamaño de muestra: {n}")
print(f"Grados de libertad: {grados_libertad}")
print(f"Varianza muestral: {s2:.4f}")
print(f"Chi-cuadrado inferior: {chi_inferior:.4f}")
print(f"Chi-cuadrado superior: {chi_superior:.4f}")
print(f"Intervalo de confianza del 95% para sigma^2:")
print(f"({limite_inferior:.2f}, {limite_superior:.2f}) lb^2")
```
